```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

# 1. Criando um dataset sintético para forçar o cenário da questão
# Atributo <= 0.5 irá para a esquerda, Atributo > 0.5 irá para a direita
X = [[0], [0], [1], [1]]

# As classes correspondentes (3 'G's e 1 'D')
y = ['G', 'G', 'G', 'D']

# 2. Configurando o modelo
# criterion='entropy' garante o uso da fórmula de Shannon (H) vista na questão original
# max_depth=1 garante que a árvore pare logo no primeiro nível de filhos
modelo = DecisionTreeClassifier(criterion='entropy', max_depth=1, random_state=42)
modelo.fit(X, y)

# 3. Gerando o gráfico no estilo visual solicitado
plt.figure(figsize=(8, 6))

plot_tree(
    modelo, 
    filled=True,                   # Pinta os nós com as cores das classes
    rounded=True,                  # Arredonda as bordas das caixas
    feature_names=['Condição X'],  # Nome fictício da regra de corte
    class_names=['D', 'G'],        # Classes em ordem alfabética
    fontsize=12                    # Tamanho da fonte
)

plt.title("Árvore de Decisão: Cenário {G, G, G, D}")
plt.show()
```
